```python

%%writefile app.py
import streamlit as st
import pandas as pd
import matplotlib.pyplot as plt

st.set_page_config(
    page_title="Quản lý điểm sinh viên",
    page_icon="📚",
    layout="wide"
)

st.title("QUẢN LÝ ĐIỂM SINH VIÊN")

# Du lieu 10 sinh vien
data = [
    ["Nguyễn Minh Anh", 8, 7, 9],
    ["Trần Quốc Bảo", 9, 8, 7],
    ["Lê Thị Cẩm", 7, 6, 8],
    ["Phạm Hoàng Duy", 10, 9, 9],
    ["Võ Ngọc Hà", 8, 8, 10],
    ["Đặng Thành Long", 6, 7, 7],
    ["Bùi Khánh Linh", 9, 9, 8],
    ["Ngô Tuấn Kiệt", 7, 8, 9],
    ["Đỗ Phương Mai", 10, 8, 9],
    ["Hoàng Gia Nam", 8, 7, 8]
]

df = pd.DataFrame(data, columns=[
    "Họ tên", "Chuyên cần", "Giữa kỳ", "Cuối kỳ"
])

# Tinh diem tong ket
df["Tổng kết"] = (
    df["Chuyên cần"] * 0.2
    + df["Giữa kỳ"] * 0.3
    + df["Cuối kỳ"] * 0.5
)

# Xep loai
def xep_loai(diem):
    if diem >= 8.5:
        return "Giỏi"
    elif diem >= 7:
        return "Khá"
    elif diem >= 5:
        return "Trung bình"
    return "Yếu"

df["Xếp loại"] = df["Tổng kết"].apply(xep_loai)

# Bang diem
st.subheader("1. Bảng điểm 10 sinh viên")
st.dataframe(df, hide_index=True, use_container_width=True)

# Thong ke
st.subheader("2. Thống kê kết quả học tập")

diem_tb = df["Tổng kết"].mean()
diem_max = df["Tổng kết"].max()
diem_min = df["Tổng kết"].min()
so_dat = (df["Tổng kết"] >= 5).sum()

c1, c2, c3, c4 = st.columns(4)
c1.metric("Điểm trung bình", f"{diem_tb:.2f}")
c2.metric("Điểm cao nhất", f"{diem_max:.1f}")
c3.metric("Điểm thấp nhất", f"{diem_min:.1f}")
c4.metric("Số sinh viên đạt", f"{so_dat}/{len(df)}")

st.write("**Sinh viên có điểm cao nhất**")
st.dataframe(
    df.loc[df["Tổng kết"] == diem_max, ["Họ tên", "Tổng kết"]],
    hide_index=True
)

st.write("**Sinh viên có điểm thấp nhất**")
st.dataframe(
    df.loc[df["Tổng kết"] == diem_min, ["Họ tên", "Tổng kết"]],
    hide_index=True
)

# Chon sinh vien
st.subheader("3. Tra cứu điểm sinh viên")
ten = st.selectbox("Chọn sinh viên", df["Họ tên"])
sv = df[df["Họ tên"] == ten].iloc[0]

a, b, c, d, e = st.columns(5)
a.metric("Chuyên cần", f"{sv['Chuyên cần']:.1f}")
b.metric("Giữa kỳ", f"{sv['Giữa kỳ']:.1f}")
c.metric("Cuối kỳ", f"{sv['Cuối kỳ']:.1f}")
d.metric("Tổng kết", f"{sv['Tổng kết']:.2f}")
e.metric("Xếp loại", sv["Xếp loại"])

# Bieu do cot
st.subheader("4. Biểu đồ điểm tổng kết")

fig, ax = plt.subplots(figsize=(12, 6))
ax.bar(df["Họ tên"], df["Tổng kết"])
ax.set_title("BIỂU ĐỒ ĐIỂM TỔNG KẾT CỦA 10 SINH VIÊN")
ax.set_xlabel("Họ tên sinh viên")
ax.set_ylabel("Điểm tổng kết")
ax.set_ylim(0, 10)
plt.setp(ax.get_xticklabels(), rotation=45, ha="right")

for i, diem in enumerate(df["Tổng kết"]):
    ax.text(i, diem + 0.1, f"{diem:.1f}", ha="center")

fig.tight_layout()
st.pyplot(fig)
plt.close(fig)

# Thong tin nguoi tao
st.divider()
st.caption("Người tạo ứng dụng: Nguyễn Hoàng Ngọc Dung")
st.caption("MSSV: 038308017227")
```
